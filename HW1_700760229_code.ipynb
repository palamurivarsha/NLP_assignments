```python
"""
Q5.1 - Naive vs. Manually-Corrected Tokenization
--------------------------------------------------
1. Naive space-based tokenization (just str.split()).
2. Manual correction: punctuation split off, clitics/contractions
   separated, hyphenated compounds kept together.
3. Diff between the two, so differences are easy to see/highlight.
"""

import difflib

# ---- 1. Your paragraph goes here ----
paragraph = (
  "ఈరోజు మన నగరంలో ఒక కొత్త గ్రంథాలయం ప్రారంభమైంది.ఈ ఆధునిక గ్రంథాలయంలో విద్యార్థుల కోసం వేలాది పుస్తకాలు అందుబాటులో ఉన్నాయి.యువత ఈ సదుపాయాన్ని ఉపయోగించుకుని మంచి పరిజ్ఞానం పొందాలని అధికారులు కోరారు."
)

# ---- 2. Naive space-based tokenization ----
naive_tokens = paragraph.split()

# ---- 3. Manually corrected tokens ----
# Write these out by hand: split off punctuation, split clitics
# (can't -> ca n't, it's -> it 's, York's -> York 's, Let's -> Let 's),
# but keep true fixed compounds like "well-known" as one token.
manual_tokens = [
    "ఈరోజు", "మన ","నగరంలో","ఒక", "కొత్త ","గ్రంథాలయం ", "ప్రారంభమైంది",".","ఈ" ,"ఆధునిక" , "గ్రంథాలయంలో ", "విద్యార్థుల","కోసం", "వేలాది ","పుస్తకాలు" , "అందుబాటులో" , "ఉన్నాయి", ".",
     " యువత", "ఈ", "సదుపాయాన్ని","ఉపయోగించుకుని", "మంచి","పరిజ్ఞానం", "పొందాలని","అధికారులు","కోరారు", "."
]

# ---- 4. Print both versions ----
print("NAIVE TOKENIZATION (%d tokens):" % len(naive_tokens))
print(naive_tokens)
print()

print("MANUAL TOKENIZATION (%d tokens):" % len(manual_tokens))
print(manual_tokens)
print()

# ---- 5. Highlight the differences ----
print("DIFFERENCES (naive -> manual):")
sm = difflib.SequenceMatcher(None, naive_tokens, manual_tokens)
for tag, i1, i2, j1, j2 in sm.get_opcodes():
    if tag == "equal":
        continue
    naive_chunk = naive_tokens[i1:i2]
    manual_chunk = manual_tokens[j1:j2]
    print(f"  {tag.upper():7} naive={naive_chunk}  ->  manual={manual_chunk}")
```

## 2.2 — Code a mini-BPE learner 

```python
"""
Q2.2 - Mini BPE Learner
------------------------
1. Learn BPE merges on the toy corpus, printing the top pair
   and the evolving vocabulary size at every step.
2. Use the learned merges to segment new words, including
   an invented word.
"""

import re
import collections

# ---- Toy corpus: word -> frequency ----
corpus = {
    "low": 5,
    "lowest": 2,
    "newer": 6,
    "wider": 3,
    "new": 2,
}


def get_vocab(corpus):
    """Turn each word into a space-separated sequence of characters
    plus an end-of-word marker '_'."""
    vocab = collections.defaultdict(int)
    for word, freq in corpus.items():
        vocab[" ".join(list(word)) + " _"] += freq
    return vocab


def get_stats(vocab):
    """Count every adjacent symbol pair across the vocab, weighted
    by word frequency."""
    pairs = collections.defaultdict(int)
    for word, freq in vocab.items():
        symbols = word.split()
        for i in range(len(symbols) - 1):
            pairs[(symbols[i], symbols[i + 1])] += freq
    return pairs


def merge_vocab(pair, vocab):
    """Merge every occurrence of `pair` into a single symbol."""
    out = {}
    bigram = re.escape(" ".join(pair))
    pattern = re.compile(r"(?<!\S)" + bigram + r"(?!\S)")
    for word in vocab:
        out[pattern.sub("".join(pair), word)] = vocab[word]
    return out


def learn_bpe(corpus, num_merges):
    """Run BPE, printing the winning pair + vocab size each step.
    Returns the ordered list of merges learned."""
    vocab = get_vocab(corpus)
    merges = []

    all_tokens = set()
    for w in vocab:
        all_tokens.update(w.split())
    print(f"Initial vocab size: {len(all_tokens)}  {sorted(all_tokens)}\n")

    for step in range(1, num_merges + 1):
        pairs = get_stats(vocab)
        if not pairs:
            break
        best = max(pairs, key=pairs.get)
        vocab = merge_vocab(best, vocab)
        merges.append(best)

        all_tokens = set()
        for w in vocab:
            all_tokens.update(w.split())

        print(f"Step {step}: merge {best} -> '{''.join(best)}'  "
              f"(count={pairs[best]})  | vocab size={len(all_tokens)}")

    return merges


def segment(word, merges):
    """Apply the learned merges, in order, to a new word."""
    symbols = list(word) + ["_"]
    for pair in merges:
        i = 0
        new_symbols = []
        while i < len(symbols):
            if (i < len(symbols) - 1
                    and symbols[i] == pair[0]
                    and symbols[i + 1] == pair[1]):
                new_symbols.append(symbols[i] + symbols[i + 1])
                i += 2
            else:
                new_symbols.append(symbols[i])
                i += 1
        symbols = new_symbols
    return symbols


if __name__ == "__main__":
    # ---- 1. Learn merges ----
    merges = learn_bpe(corpus, num_merges=10)

    # ---- 2. Segment target words (including an invented one) ----
    test_words = ["new", "newer", "lowest", "widest", "newestest"]

    print("\nSegmentation results:")
    for w in test_words:
        print(f"  {w:12} -> {segment(w, merges)}")
```

### 2.3 train BPE

```python
from collections import Counter
import re

text_block = """
Natural language processing helps computers understand human language.
Tokenization separates words, punctuation, and contractions.
Students learn from examples and compare patterns across sentences.
Thoughtful preprocessing improves reliable models for practical language tasks.
Reusable subwords help learners analyze uncommon words and inflected forms.
"""

# Extract lowercase alphabetic words and append an end-of-word marker
word_list = re.findall(r"[a-z]+", text_block.lower())
token_vocab = Counter(tuple(term) + ("_",) for term in word_list)

starting_symbols = set()
for token_sequence in token_vocab:
    starting_symbols.update(token_sequence)

merge_history = []


def count_pairs(current_vocab):
    pair_frequency = Counter()

    for sequence, occurrence in current_vocab.items():
        for adjacent_pair in zip(sequence, sequence[1:]):
            pair_frequency[adjacent_pair] += occurrence

    return pair_frequency


def apply_merge(target_pair, current_vocab):
    updated_vocab = Counter()
    combined_token = "".join(target_pair)

    for sequence, occurrence in current_vocab.items():
        rebuilt_sequence = []
        position = 0

        while position < len(sequence):
            if (
                position < len(sequence) - 1
                and (sequence[position], sequence[position + 1]) == target_pair
            ):
                rebuilt_sequence.append(combined_token)
                position += 2
            else:
                rebuilt_sequence.append(sequence[position])
                position += 1

        updated_vocab[tuple(rebuilt_sequence)] += occurrence

    return updated_vocab


# Learn 30 BPE merges
for iteration in range(30):
    pair_stats = count_pairs(token_vocab)

    if not pair_stats:
        break

    selected_pair = max(pair_stats, key=pair_stats.get)
    pair_count = pair_stats[selected_pair]
    merged_symbol = "".join(selected_pair)

    merge_history.append({
        "pair": selected_pair,
        "token": merged_symbol,
        "frequency": pair_count
    })

    token_vocab = apply_merge(selected_pair, token_vocab)

    print(
        f"Step {iteration + 1}: "
        f"{selected_pair[0]} + {selected_pair[1]} -> {merged_symbol} "
        f"(frequency = {pair_count})"
    )


# Display the five most frequent merges
most_common_merges = sorted(
    merge_history,
    key=lambda record: record["frequency"],
    reverse=True
)[:5]

print("\nFive most frequent merges:")
for record in most_common_merges:
    print(
        f"{record['pair'][0]} + {record['pair'][1]} "
        f"-> {record['token']} "
        f"(frequency = {record['frequency']})"
    )


# Display the five longest learned subword tokens
known_tokens = starting_symbols | {
    record["token"] for record in merge_history
}

longest_subwords = sorted(
    known_tokens,
    key=lambda symbol: (-len(symbol), symbol)
)[:5]

print("\nFive longest subword tokens:")
for symbol in longest_subwords:
    print(symbol)


def tokenize_word(input_word, merge_rules):
    current_symbols = list(input_word.lower()) + ["_"]

    for rule in merge_rules:
        target = rule["pair"]
        replacement = rule["token"]

        transformed = []
        position = 0

        while position < len(current_symbols):
            if (
                position < len(current_symbols) - 1
                and (current_symbols[position], current_symbols[position + 1]) == target
            ):
                transformed.append(replacement)
                position += 2
            else:
                transformed.append(current_symbols[position])
                position += 1

        current_symbols = transformed

    return current_symbols


# Segment five words from the paragraph
sample_words = [
    "language",
    "tokenization",
    "uncommon",
    "inflected",
    "learners"
]

print("\nWord segmentations:")
for sample in sample_words:
    pieces = tokenize_word(sample, merge_history)
    print(f"{sample} -> {' | '.join(pieces)}")
```
